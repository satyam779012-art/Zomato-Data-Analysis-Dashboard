```python
import pandas as pd

df = pd.read_csv("Zomato.csv")
print(df.head())
```

```python
df
```

```python
## Convert the Data type of Coloum - Rate##
```

```python
def handleRate(value):
    value = str(value).split("/")
    return float(value[0])

df["rate"] = df["rate"].apply(handleRate)

print(df.head())
```

```python
df
```

```python
df.info()
```

```python
## What type of restaurant do the majority of customers order from? ##
```

```python
df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

ax = sns.countplot(x=df['listed_in(type)'])

colors = ['red', 'blue', 'orange', 'green']

for bar, color in zip(ax.patches, colors):
    bar.set_color(color)

plt.title("Show the countplot",color='red',size=15)
plt.xlabel("Type of Restaurant",color='Green',size=15)
plt.ylabel("Count",color='orange',size=15)


plt.show()
```

```python
 ## Q2. How many votes has each type of restaurant received from customers? ##
```

```python
df.head()
```

```python
from matplotlib.lines import lineStyles
group_data=df.groupby(['listed_in(type)'])['votes'].sum()
result=pd.DataFrame({'votes':group_data})
plt.plot(result,color='red',marker='o')
plt.title("Show the Plot Line",c='orange',size=20)
plt.xlabel("Type of Restaurant",color='red',size=20)
plt.ylabel("Votes",color='Green',size=20)
plt.grid(True)
plt.show
```

```python
## Q3. What are the ratings that the majority of restaurants have received? ##
```

```python
plt.hist(df['rate'],bins=20,color='orange')
plt.title("Ratings Distbution",color='red',size=15)
plt.xlabel("Ratings",color='green',size=10)
plt.ylabel("Count",color='blue',size=10)
plt.show()
```

```python
## Q4.)Zomato has observed that most couples order most of their food online. What is their average spending on each order? ##
```

```python
df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

ax = sns.countplot(x=df['approx_cost(for two people)'])

colors = ['red', 'blue', 'green', 'orange', 'purple',
          'cyan', 'yellow', 'pink', 'brown', 'gray']

for i, bar in enumerate(ax.patches):
    bar.set_color(colors[i % len(colors)])

plt.title("Show the Couples Chart",color='cyan')
plt.xlabel("Value", color='green', size=10)
plt.ylabel("Count", color='blue', size=10)

plt.show()
```

```python
## Q5.Which mode (online or offline) has received the maximum rating? ##
```

```python
df.head()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    x='online_order',
    y='rate',
    data=df
)

plt.title("Show the maximum rating",color='red',size=15)
plt.xlabel("Online Order",color='green')
plt.ylabel("Rate",color='Pink')

plt.show()
```

```python
## Q6. Which type of restaurant received more offline orders, so that Zomato can provide those customers with some good offers? ##
```

```python
df.head()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

pivot_table = df.pivot_table(
    index='listed_in(type)',
    columns='online_order',
    aggfunc='size',
    fill_value=0
)

plt.figure(figsize=(8,5))

sns.heatmap(
    pivot_table,
    annot=True,
    cmap="YlGnBu",
    fmt='d'
)

plt.title("Heatmap")
plt.xlabel("Online Order")
plt.ylabel("Listed In (Type)")

plt.show()
```
